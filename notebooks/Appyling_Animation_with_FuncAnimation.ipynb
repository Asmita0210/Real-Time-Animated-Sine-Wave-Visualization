```python
import matplotlib.pyplot as plt
import matplotlib.animation as animation
import numpy as np
```

```python
fig, ax = plt.subplots(figsize=(9, 4))
ax.set_xlim(0, 4*np.pi)
ax.set_ylim(-1.5, 1.5)
ax.set_title('Animated Sine wave')
ax.grid(True, alpha=0.3)

line, = ax.plot([], [], color='#2563A8', linewidth=2)
x_data, y_data = [], []

def  init():
    line.set_data([], [])
    return line,

def update(frame):
    x = frame * 0.05
    x_data.append(x)
    y_data.append(np.sin(x))

    if len(x_data) > 200:
        x_data.pop(0); y_data.pop(0)
        ax.set_xlim(x_data[0], x_data[0] + 4*np.pi)
    line.set_data(x_data, y_data)
    return line,

ani = animation.FuncAnimation(
    fig, update, init_func = init,
    frames=1000, interval=20,
    blit=True
)

#Save animation
ani.save('sine_wave.gif', writer='pillow', fps=30)
plt.show()
```
